Desarrolla un programa que permita registrar el consumo eléctrico mensual de 10 viviendas, expresado en kilovatios hora. El sistema debe validar que cada consumo sea mayor que 0. Luego, debe clasificar cada vivienda según su consumo: bajo si consume hasta 100 kWh, moderado si consume más de 100 y hasta 250 kWh, y alto si supera los 250 kWh. Además, si una vivienda tiene consumo alto y supera los 400 kWh, debe marcarse como consumo crítico. Al finalizar, el programa debe mostrar cuántas viviendas pertenecen a cada categoría y cuántas requieren revisión por consumo crítico.

```python
viviendas_crit = 0
con_bajo = 0
con_medio = 0
con_alto = 0

for i in range(1,11):
    while True:
        try:
            consumo_str = input(f"Ingrese el consumo de la vivienda {i}: ")
            consumo = float(consumo_str)
            if consumo <= 0:
                print("El consumo debe ser mayor que 0. Por favor, intente de nuevo.")
            else:
                break # Exit loop if input is valid and positive
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número para el consumo.")

    if consumo <= 100:
        con_bajo += 1
    elif consumo <= 250:
        con_medio += 1
    else:
        con_alto += 1
        if consumo > 400:
            viviendas_crit += 1

print(f"Viviendas con consumo bajo: {con_bajo}")
print(f"Viviendas con consumo medio: {con_medio}")
print(f"Viviendas con consumo alto: {con_alto}")
print(f"Cantidad de viviendas que requieren revision por consumo critica: {viviendas_crit}")
```
